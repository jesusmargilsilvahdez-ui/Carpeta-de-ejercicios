```python
# Función para calcular la velocidad
def calcular_velocidad(distancia, tiempo):
    if tiempo == 0:
        return "El tiempo no puede ser cero"
    else:
        velocidad = distancia / tiempo
        return f"La velocidad es: {velocidad} m/s"

# Solicitar datos
distancia = float(input("Ingrese la distancia (en metros): "))
tiempo = float(input("Ingrese el tiempo (en segundos): "))

# Calcular y mostrar el resultado
resultado = calcular_velocidad(distancia, tiempo)
print(resultado)
```

```python
import math

# Función para calcular el MCM
def calcular_mcm(a, b):
    mcd = math.gcd(a, b)
    mcm = abs(a * b) // mcd
    return mcm

# Solicitar los números
a = int(input("Ingrese el primer número: "))
b = int(input("Ingrese el segundo número: "))

# Calcular y mostrar el resultado
resultado = calcular_mcm(a, b)
print(f"El MCM de {a} y {b} es: {resultado}")
```
